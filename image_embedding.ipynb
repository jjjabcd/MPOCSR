```python
import os
import sys
sys.path.append(os.path.abspath('model/MPViT-CBloss'))


import pandas as pd
import numpy as np

import torch
from pre_transformer import Encoder
```

```python
class Args:
    def __init__(self, model_name='mpvit_base', tag=True):
        self.model_name = model_name
        self.tag = tag
```

```python
model_path = 'ckpt/MPViT_CBLoss.pth'
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
args = Args()
encoder = Encoder(args.model_name, args.tag).to(device)
encoder.load_state_dict(torch.load(model_path, map_location=device))
encoder.eval()
```
